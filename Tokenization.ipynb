```python
corpus="""Hi my name is Ashish kumar's  Rastogi.
Please tell me what is your name. he is a good teacher."""
print(corpus)
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
# Tokenization
#from Sentence to paragraph
from nltk.tokenize import sent_tokenize #it's try to convert a paragraph in sentences

```

```python
document=sent_tokenize(corpus)
```

```python
arr=[]
for sentence in document:
  arr.append(sentence)

print(arr)
```

```python
#Sentence to word
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
for sentence in document:
  print(sentence)
  print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tree_word=TreebankWordTokenizer()
```

```python
tree_word.tokenize(corpus)
```
